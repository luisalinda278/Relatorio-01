## 🥈 Exercício 3 — A régua do tempo: quanto o seu município cresceu de fato

O kit te deu **dois pontos** da série do PIB per capita: `MEU_PIB_PC_BASE` no
ano `MEU_ANO_BASE`, e `MEU_PIB_PC` no ano `MEU_ANO`. A pergunta do relatório é
a do Desafio 1 da Missão 1: **qual foi o crescimento médio por ano?**

E a resposta **não** é "crescimento total ÷ número de anos" — porque o
crescimento se capitaliza. A taxa média geométrica (*CAGR*) é:

$$\text{CAGR} = \left(\frac{V_{final}}{V_{inicial}}\right)^{1/n} - 1$$

> ⚠️ Conte os **intervalos**, não os anos: de 2019 a 2023 são **4** períodos
> de crescimento, não 5. É o erro clássico, e ele contamina o CAGR inteiro.

> 💡 O seu crescimento pode ter sido **negativo** — há município brasileiro
> que encolheu no período, e a fórmula lida com isso naturalmente. Se der
> negativo, não é bug: é economia.

No fim, a comparação que interessa ao gestor: o plano diretor projeta
`MEU_CRESCIMENTO`% ao ano daqui para a frente. Isso é mais otimista do que o
município de fato entregou?

````python
# TODO 1: quantos ANOS de crescimento separam os dois pontos da série
meu_ano_base = float(2016)
meu_ano = float(2021)
pib_per_capita = float(25865.93) 
pib_ano_base = float(26789.95)  
anos_da_serie  = (meu_ano - meu_ano_base)
crescimento = float(0.00308)
municipio = str("Fortaleza")

# TODO 2: o crescimento total do período, em %
crescimento_total_pct= ((pib_per_capita / pib_ano_base)**(1)-1)*100

# TODO 3: o CAGR, em % ao ano (dica: expoente fracionário com **)
cagr_pct = ((pib_per_capita / pib_ano_base)**(1/anos_da_serie)-1)*100

# TODO 4: a média aritmética "ingênua" que muita gente reporta por engano
media_ingenua_pct = (crescimento_total_pct / anos_da_serie)


# TODO 5: um bool, sem if — a projeção do plano diretor (MEU_CRESCIMENTO)
# é mais otimista que o CAGR que o município realmente entregou?
projecao_otimista = True if crescimento > cagr_pct else False

print(f" {municipio}, de {meu_ano_base} a {meu_ano} ({anos_da_serie} anos):")
print(f"  PIB per capita de {municipio} e R$ {pib_ano_base:.2f} -> R$ {pib_per_capita:.2f}")
print(f"  crescimento total de {municipio} {crescimento_total_pct:.2f}%")
print(f"  CAGR (correto) de {municipio} {cagr_pct:.2f}% ao ano")
print(f"  média ingênua (errada) {media_ingenua_pct:.2f}% ao ano")
print(f"  projeção do plano diretor: {crescimento}% ao ano")
print("  Plano diretor mais otimista que o histórico?", projecao_otimista)
````

````python
conferir(
    "exercicio-3",
    anos_da_serie=anos_da_serie,
    crescimento_total_pct=crescimento_total_pct,
    cagr_pct=cagr_pct,
    media_ingenua_pct=media_ingenua_pct,
    projecao_otimista=projecao_otimista,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-3", "fazendo a Taxa medida geometrica CAGR tive problema com o kernel, precisei reiniciar o kernei e rodar novamente.")
````

## 🥈 Exercício 4 — A régua do poder de compra: a equação de Fisher

A prefeitura aplicou o superávit de caixa e rendeu `MINHA_TAXA_NOMINAL`% no
ano. No mesmo período, a inflação foi de `MINHA_INFLACAO`%. **O município
ficou mais rico?**

O ganho **real** não é a subtração simples: a fórmula correta é a **equação de
Fisher exata**, a mesma da Prática 2 da Missão 1.

$$1 + r = \frac{1 + i}{1 + \pi} \qquad\Longrightarrow\qquad
r = \frac{1 + i}{1 + \pi} - 1$$

onde $r$ = taxa real, $i$ = taxa nominal, $\pi$ = inflação — todas em
**fração**, não em percentual. As suas duas variáveis chegam em percentual;
converter é com você.

Calcule as duas versões — a exata e a aproximação de bolso ($i - \pi$) — e
mostre em quantos **pontos percentuais** elas discordam. Depois, o semáforo do
Aquecimento 1 da Missão 2:

| Situação | Sinal |
|---|---|
| $r > 2\%$ | 🟢 ganho real relevante |
| $0 < r \le 2\%$ | 🟡 ganho real modesto |
| $r \le 0$ | 🔴 perda de poder de compra |

````python
taxa_nominal = float(10.93)
inflacao = float(8.47)

# TODO 1: Fisher exata, resultado em PERCENTUAL
#         (converta as duas taxas para fração antes; multiplique por 100 depois)
taxa_real_pct = ((1+(taxa_nominal / 100))/(1+(inflacao / 100)) -1) * 100

# TODO 2: a aproximação de bolso, em percentual
taxa_real_aproximada_pct = (taxa_nominal - inflacao)

# TODO 3: a diferença, em pontos percentuais — quanto a aproximação
#         SUPERESTIMA o ganho real. O sinal importa.
erro_pp = taxa_real_aproximada_pct - taxa_real_pct 

# TODO 4: o semáforo, com if/elif/else
if erro_pp > 0.02:
    sinal_fisher = "🟢 ganho real relevante"
elif erro_pp <= 0.02 and erro_pp > 0:
    sinal_fisher = "🟡 ganho real modesto"
else:
    sinal_fisher = "🔴 perda de poder de compra"

print(f"Nominal: {taxa_nominal * 100:.2f}%  |  Inflação: {inflacao * 100:.2f}%")
print(f"  taxa real (Fisher exata) .. {taxa_real_pct:.2f}%")
print(f"  taxa real (aproximação) ... {taxa_real_aproximada_pct:.2f}%")
print(f"  a aproximação superestima em {erro_pp:.2f} p.p.")
print(f"Parecer sobre a aplicação: {sinal_fisher}")
````

````python
conferir(
    "exercicio-4",
    taxa_real_pct=taxa_real_pct,
    taxa_real_aproximada_pct=taxa_real_aproximada_pct,
    erro_pp=erro_pp,
    sinal_fisher=sinal_fisher,
)

registrar("exercicio-4", "eu primeiro fiz a equação de forma inversa:  - taxa_real_pct - taxa_real_aproximada_pct, depois mudein e diu certo")

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
````

## 🥈 Exercício 5 — A régua da meta: a projeção até 2030

O plano diretor do seu município fixou uma meta de PIB per capita para
**2030** — é a variável `MINHA_META`, em R$ por habitante. Partindo do seu ano
foco (`MEU_ANO`) e do crescimento projetado (`MEU_CRESCIMENTO`), a meta é
alcançável?

A projeção usa a mesma fórmula dos juros compostos do Aquecimento 1 da
Missão 1:

$$V_{2030} = V_{\text{ano foco}} \times \left(1 + \frac{g}{100}\right)^{h}$$

onde $h$ é o **horizonte**, em anos, entre o ano foco e 2030 — contado como no
Exercício 3.

Depois da projeção, uma cadeia de três faixas para o parecer:

| Situação | Parecer |
|---|---|
| projeção ≥ meta | ✅ meta alcançada |
| entre 90% da meta e a meta | ⚠️ meta próxima, exige aceleração |
| abaixo de 90% da meta | ❌ meta inalcançável no ritmo atual |

````python
meu_ano = float(2021)
crescimento = float(3.08)
meta = float(35300.00)  # R$ 35300.00 (PIB per capita em 2030)
pib_per_capita = float(25865.93) 
municipio = str("Fortaleza")

# TODO 1: o horizonte, em anos, entre o seu ano foco e 2030
horizonte = 2030 - meu_ano

# TODO 2: a projeção (atenção aos parênteses — lembre da Parte 1!)
pib_projetado = pib_per_capita * ((1 +  (crescimento/100) )** horizonte)

# TODO 3: quanto isso representa da meta, em %
percentual_da_meta = (pib_projetado / meta) * 100

# TODO 4: o parecer, com if/elif/else
if pib_projetado >= meta:
    parecer = "✅ meta alcançada"
elif pib_projetado >= meta * 0.90:
    parecer = "⚠️ meta próxima, exige aceleração"
else:
    parecer = "❌ meta inalcançável no ritmo atual"

print(f"De {meu_ano} a 2030 são {horizonte} anos de crescimento.")
print(f"PIB per capita projetado: R$ {pib_projetado:.2f}")
print(f"Isso é {percentual_da_meta:.1f}% da meta de R$ {meta:.2f}")
print(f"Parecer para {municipio}: {parecer}")
````

## 🥇 Exercício 6 — A régua fiscal: alíquota marginal × alíquota efetiva

Cada município contribui para o **Fundo Regional de Desenvolvimento** com uma
alíquota que cresce por faixa de PIB per capita — exatamente o desenho da
tabela do IRPF que você viu na Prática 2 da Missão 2, com **parcela a
deduzir** e tudo:

| Faixa de PIB per capita | Alíquota | Parcela a deduzir |
|---|---|---|
| até R$ 18.000 | isento | — |
| acima de R$ 18.000 até R$ 28.000 | 1,0% | R$ 180,00 |
| acima de R$ 28.000 até R$ 45.000 | 2,5% | R$ 600,00 |
| acima de R$ 45.000 | 4,5% | R$ 1.500,00 |

$$\text{contribuição por habitante} = \text{PIB per capita} \times
\text{alíquota} - \text{parcela a deduzir}$$

E aqui mora a confusão que todo economista precisa saber desmontar em 30
segundos — **duas alíquotas convivem, e elas não são a mesma coisa**:

- a **marginal** é a da tabela, a que incide sobre o *próximo* real de PIB;
- a **efetiva** é o que o município de fato paga: contribuição ÷ PIB per
  capita.

> 💡 Repare que a alíquota e a parcela a deduzir **andam sempre juntas**: são
> duas atribuições no mesmo ramo do `if`. Trocar uma sem trocar a outra é o
> bug mais comum desta tabela.

````python
# TODO 1 e 2: a alíquota marginal E a parcela a deduzir, com if/elif/else.
# A cadeia vai da MENOR faixa para a maior, porque o operador é <=

pib_per_capita = float(25865.93) 
if pib_per_capita <= 18000:
    aliquota = 0.000
    parcela_deduzir = 0.00
elif pib_per_capita <= 28000 and pib_per_capita > 18000:
    aliquota = 0.010
    parcela_deduzir = 180.00
elif pib_per_capita <= 45000 and pib_per_capita > 28000:
    aliquota = 0.025
    parcela_deduzir = 600.00
else:
    aliquota = 0.045
    parcela_deduzir = 1500.00

# TODO 3: a contribuição por habitante, em R$
contribuicao_por_habitante = pib_per_capita * aliquota - parcela_deduzir

# TODO 4: a contribuição total do município, em mil R$
#         (população em mil hab. × R$ por hab. = mil R$)
contribuicao_total = populacao * contribuicao_por_habitante

# TODO 5: a alíquota EFETIVA, em % — o que o município realmente paga
aliquota_efetiva_pct = contribuicao_total / pib_total * 100

# TODO 6: um bool combinando DUAS condições com `and`, sem nenhum if.
# O município é contribuinte relevante se paga alíquota marginal acima de
# 1,0% E contribui com mais de R$ 50.000 mil ao fundo.
contribuinte_relevante = aliquota > 0.01 and contribuicao_total > 50000

print(f"{municipio}: PIB per capita R$ {pib_per_capita:.2f}")
print(f"  alíquota marginal ..... {aliquota * 100:.2f}%")
print(f"  parcela a deduzir ..... R$ {parcela_deduzir:.2f}")
print(f"  por habitante ......... R$ {contribuicao_por_habitante:.2f}")
print(f"  contribuição total .... R$ {contribuicao_total:.2f} mil")
print(f"  alíquota EFETIVA ...... {aliquota_efetiva_pct:.2f}%")
print("Contribuinte relevante?", contribuinte_relevante)
````

## 🥇 Exercício 7 — O semáforo fiscal blindado

Este é o exercício que junta tudo o que a Missão 2 apresentou.

A dívida consolidada do seu município chega do portal da transparência **como
texto** — a variável `MINHA_DIVIDA` é uma `str`, como você viu no Passo 0. E
dado público chega sujo: pode vir `"n/d"`, pode vir vazio, pode vir com um
valor economicamente impossível.

Sua rotina precisa sobreviver a **dois tipos de problema diferentes**, e cada
um pede uma ferramenta:

| Problema | Exemplo | Ferramenta |
|---|---|---|
| Dado ilegível | `"n/d"`, `""` | `try` / `except ValueError` |
| Regra de negócio violada | `"-5"`, `"450"` | `raise ValueError` |

A regra de negócio: uma dívida em percentual da receita fora do intervalo
**[0, 300]** é dado corrompido, não município endividado.

E o semáforo, para o dado que passar na triagem:

| Situação | Sinal |
|---|---|
| dívida ≤ 40% | 🟢 risco baixo |
| acima de 40% até 60% | 🟡 risco moderado |
| acima de 60% | 🔴 risco alto |

> 💡 O padrão é o da Missão 2: o `try`/`except` responde *"o dado existe e é
> legítimo?"*; a cadeia `if`/`elif`/`else`, **dentro do `try`**, responde *"em
> que faixa ele cai?"*. Um único `except ValueError` cobre os dois problemas,
> porque o seu `raise` produz o mesmo tipo de exceção que o `float()` gera.

````python
# O dado como ele chega do portal — texto, não número
divida_texto = 23.9
divida_pct =   23.9

try:
    # TODO 1: converta o texto para número
    divida_pct = float(divida_texto)

    # TODO 2: a regra de negócio — levante ValueError com mensagem clara
    if divida_pct == "n/d" or divida_pct == "":
        raise ValueError("Dado ilegível")

    # TODO 3: o semáforo, com if/elif/else
    if divida_pct <= 40:
        sinal = "🟢 risco baixo"
    elif divida_pct > 40 and divida_pct <= 60:
        sinal = "🟡 risco moderado"
    else:
        sinal = "🔴 risco alto"

except ValueError as erro:
    # TODO 4: exiba a mensagem capturada e marque o caso como sem dado
    sinal = "⚪ sem dado"
    print(f"⚠️ Triagem: {erro}")

print(f"{municipio} | dívida informada: {divida_texto!r} -> {sinal}")
````
